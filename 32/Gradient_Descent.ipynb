## making linear regression dataset  

```python
from sklearn.datasets import make_regression #It is used to create random linear regression datasets with controlled noise, number of features, informative features, and more
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y = make_regression(n_samples=4 , n_features=1 , n_informative=1  ,n_targets=1 , noise = 40 , random_state=13)
```

```python
plt.scatter(X,y)
```

### Applying OLS(ordinary least square)

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(X,y)
```

#### checking intercept and slope of linear regression line

```python
reg.coef_
```

```python
reg.intercept_
```

#### Plotting the best fit line

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color = 'red')
```

## Apllying Gradient Descent assuming slope is constant m = 69.6274

#### Also assuming the starting value for intercept b = 0

```python
y_pred = ((69.6274 * X) + 100).reshape(4)
```

```python
y_pred
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X) , color= 'red' , label = 'OLS')
plt.plot(X,y_pred , color= '#00a65a' , label = 'b = 0')
plt.legend()
plt.show()
```

#### Now iteratively finding the new b until it converge with the best fit line

```python
m = 69.6274
b = 0

loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

#### Taking learning rate to be 0.1

```python
lr = 0.1
step_size = loss_slope*lr
step_size
```

```python
#calculating the new intercept 
b = b - step_size
b
```

```python
y_pred1 = ((69.6274 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
```

```python
# iteration2 
#first calculating the slope
loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
# step2 -> calculating the step size by multiplying learning rate and loss slope
step_size = loss_slope*lr
step_size
```

```python
# step 3 -> calculating the new b
b = b - step_size
b
```

```python
#Again plotting to see the progress
y_pred2 = ((69.6274 * X) + b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X) , color = 'red' , label = 'OLS')
plt.plot(X,y_pred2 , color = '#00a65a' , label = 'b = {}'.format(b))
plt.plot(X,y_pred1 , color = '#A3E4D7' , label = 'b = 43.46')
plt.plot(X,y_pred , color = '#A3E4D7' , label = 'b = 0')
plt.legend()
plt.show()
```

```python
# Iteration 3
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((69.6274* X) + b).reshape(4)

plt.figure(figsize=(15,15))
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred3,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred2,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X,y_pred1,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
     
```

### Now visualization with the help of a loop with a changed b and a epochs value

```python
b = -100
m = 69.6274
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```
