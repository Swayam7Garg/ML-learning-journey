```python
from sklearn.datasets import make_classification
import numpy as np
X , y  = make_classification(n_samples=100,n_features=2,n_informative=1 , n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=41,hypercube = False , class_sep= 10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0] , X[:,1],c=y,cmap = 'winter',s = 100)
```

```python
def perceptron(X,y):

    X = np.insert(X , 0 ,1 ,axis = 1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(1,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j] - y_hat)*X[j]

    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
def func(X,y):

    X = np.insert(X , 0 ,1 ,axis = 1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(1,100)
        y_hat = sigmoid(np.dot(X[j],weights))
        weights = weights + lr*(y[j] - y_hat)*X[j]

    return weights[0],weights[1:]
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
intercept_ , coef_ = perceptron(X,y)
```

```python
coef_.shape
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
intercept_1 , coef_1 = func(X,y)
```

```python
print(intercept_1,coef_1)# basically this is the equation Ax + By + c
```

```python
m1 = -(coef_1[0]/coef_1[1])
b1 = -(intercept_1/coef_1[1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m1*x_input1 + b1
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lgr = LogisticRegression()
lgr.fit(X,y)
```

```python
m = -(lgr.coef_[0][0]/lgr.coef_[0][1])
b = -(lgr.intercept_/lgr.coef_[0][1])
```

```python
x_input2 = np.linspace(-3,3,100)
y_input2 = m*x_input2 + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color = 'red',label = 'Perceptron')
plt.plot(x_input1,y_input1,color = 'purple',label = 'Sigmoid')
plt.plot(x_input2,y_input2,color = 'black',label = 'LogisticRegression')
plt.scatter(X[:,0] , X[:,1],c = y ,cmap='winter')
plt.ylim(-3,2)
plt.title('Perceptron vs Sigmoid vs LogisticRegression')
plt.legend()
```

### Conclusion:
- The Perceptron, Sigmoid, and Logistic Regression models all serve as fundamental building blocks in machine learning, particularly for binary classification tasks.
- The Perceptron is a simple linear classifier that updates weights based on misclassifications, making it suitable for linearly separable data.
- The Sigmoid function introduces non-linearity, allowing models to output probabilities, which is crucial for interpreting results in a probabilistic manner.
- Logistic Regression builds on the Sigmoid function to model the probability of class membership, making it a powerful tool for binary classification.
